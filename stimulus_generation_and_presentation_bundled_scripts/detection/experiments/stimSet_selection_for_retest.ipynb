# This notebook is used to generate a list of stimuli that was unused for retest

```python
# importing packages and setting up firebase 
import pandas as pd
import os
import json
import numpy as np
import random
```

```python
## load the extracted stim set
df_stimSet = pd.read_csv("./data_extraction/stimSet_dec_ses1.csv")

# import stimulus file
# load stimuli condition list as dataframe
df_stim = pd.read_csv('../all_videos.csv',index_col=[0])

# load the pre-determined stimulus set list
with open('../stimSets.js', 'r') as file:
    stimsets_all = file.read() 

stimsets_all = eval(stimsets_all)
stimsets_all = stimsets_all['file_ind']
```

```python
def get_unused_numbers(row):
    used_numbers = set(row['stim_ind_ses1'])
    unused_numbers = [num for num in unique_elements if num not in used_numbers]
    # Shuffle the order of elements
    random.shuffle(unused_numbers)
    return unused_numbers
```

```python
df_stimSet['stim_ind_ses1'] = df_stimSet['Stim_set Row_ses1'].apply(lambda x: stimsets_all[x])
df_stimSet
```

```python
stimset_all_flat = [item for sublist in stimsets_all for item in sublist]
# Get unique elements
unique_elements = list(set(stimset_all_flat))
# Apply the function to each row to get the unused numbers
df_stimSet['stim_ind_ses2'] = df_stimSet.apply(get_unused_numbers, axis=1)
df_stimSet
```

```python
stimset_ses2 = {key: value for key, value in zip(df_stimSet['Prolific ID'], df_stimSet['stim_ind_ses2'])}
stimset_ses2
output_file_path = './stimSet_generation/stimset_ses2.json'

# Write the dictionary to the JSON file
with open(output_file_path, 'w') as f:
    json.dump(stimset_ses2, f)
```
